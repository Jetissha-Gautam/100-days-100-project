```python
# @title Default title text
def encript (word,shift):
  code=""

  for i in word:
    if ord(i) not in range(97,123):
      code+=i
    elif (ord(i)+shift>122):
      #print(ord(i))
      a=(ord(i)+shift)%122+96
      #print((a))
      code+=chr(a)
    else:
      code+=chr(ord(i)+shift)
  print("Here's your encripted text:",code)
def decript(word,shift):
  code=""
  for i in word:
    if ord(i) not in range(97,123):
      code+=i
    if (ord(i)-shift<97):
      #print(ord(i))
      a=123-(97-(ord(i)-shift)%97)
      #print((a))
      code+=chr(a)
    else:
      code+=chr(ord(i)-shift)
  print("Here's your dencripted text:",code)
todo="yes"
while todo=="yes":
  print("Welcome to Caeser Cipher!!")
  print("Do you want to encode or decode?")
  task=input()
  task=task.lower()
  word=input("What's your word?")
  shift=int(input("How much is the shift?"))

  if task=='encode':
    encript(word,shift)
  elif task=="decode":
    decript(word,shift)

  else:
    print("Invalid")
  todo=input("Do you want to again encode or decode? ").lower()
else:
  print("Thankyou for using Ceaser Cipher!!!")

```
