```python
# @title Calculator
def add(n1,n2):
  return n1+n2
def sub(n1,n2):
  return n1-n2
def mul(n1,n2):
  return n1*n2
def div(n1,n2):
  return (n1/n2)
def calculation(n1):
  op=input('''What operation do you want to perform ?
  +
  -
  /
  *
  ''')
  n2=float(input("Enter the second number : "))
  print(n1 , op , n2 ,"=", operators[op](n1,n2))
  n1=operators[op](n1,n2)
  con=input(f"Do you what to continue calculating with {n1}?").lower()
  if con=="no":
    return
  calculation(n1)
operators={"+":add,"-":sub,"*":mul,"/":div}
n1=float(input("Enter the first number : "))
calculation(n1)
```
