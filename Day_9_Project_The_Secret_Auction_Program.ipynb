```python
# @title Secret Auction
from IPython.display import clear_output
more_biders=True
auction=dict()
while(more_biders==True):
  clear_output(wait=False)
  lock=r'''
     .--------.
    / .------. \
   / /        \ \
   | |        | |
  _| |________| |_
.' |_|        |_| '.
'._____ ____ _____.'
|     .'____'.     |
'.__.'.'    '.'.__.'
'.__  | YALE |  __.'
|   '.'.____.'.'   |
'.____'.____.'____.'
'.________________.'
'''
  print(lock)
  print("Welcome to The Secret Auction!!!")
  key=input("What is your name ? ")
  bid=int(input("What is your Bid ? "))
  auction[key]=bid
  more=input("Are there more bidders ?").lower()
  if more=="no":
     more_biders=False
max=0
winner=""
for i in auction:
  if auction[i]>max:
    max=auction[i]
    winner=i
print("The winner is",winner,"with a bid of",max)
```
