```python
import random
scissors=r'''
  _       ,/'
  (_).  ,/'
   _  ::    Scissors
  (_)'  `\.
           `\.'''
rock = r'''
    _________
  /         \
 |   ROCK    |
  \_________/
'''
paper = r'''
  ___________
 |           |
 |   PAPER   |
 |           |
 |___________|
'''
choices=[rock,paper,scissors]
choice=["rock","paper","scissors"]
print("Choose Rock, Paper or Scissors")
user=input().lower().strip()
user=choice.index(user)
print("You choose")
print(choices[user])
computer_choice=random.randint(0,2)
print("The computer choose" ,computer_choice)
print(choices[computer_choice])
if user==computer_choice:
  print("Its a tie!!!")
elif user==0 and computer_choice==3:
  print("You WIN!!!")
elif user>computer_choice:
  print("You WIN!!!")
else:
  print("You LOSE :(")

```
