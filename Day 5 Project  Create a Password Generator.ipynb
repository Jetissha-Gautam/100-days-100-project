```python
import random
print("Welcome to Password Generator!")
print("How many letters will you like in your password?")
alpha=int(input())
print("How many numbers will you like in your password?")
numbers=int(input())
print("How many symbols will you like in your password?")
symbols=int(input())
total=alpha+numbers+symbols
password=[]
for i in range(0,alpha):
  password+=chr(random.randint(95,112))
for i in range(0,numbers):
  password+=chr(random.randint(48,57))
for i in range(0,symbols):
  password+=chr(random.randint(33,47))
random.shuffle(password)
print("Here's your password : ", "".join(password))
```
