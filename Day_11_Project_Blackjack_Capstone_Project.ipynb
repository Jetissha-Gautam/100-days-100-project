```python
# @title BlackJack


import random
logo=r'''
                               _     _            _    _            _
                              | |   | |          | |  (_)          | |
                              | |__ | | __ _  ___| | ___  __ _  ___| | __
                              | '_ \| |/ _` |/ __| |/ / |/ _` |/ __| |/ /
                              | |_) | | (_| | (__|   <| | (_| | (__|   <
                              |_.__/|_|\__,_|\___|_|\_\ |\__,_|\___|_|\_\
                                                     _/ |
                                                    |__/

'''
You_lose=r'''
╔════════════════╗
║    YOU LOSE    ║
╚════════════════╝
'''
You_won=r'''
╔════════════════╗
║    YOU WON!    ║
╚════════════════╝
'''
print(logo)
deck=[11,2,3,4,5,6,7,8,9,10,10,10,10]
play=input("Do you want to play BlackJack??? ").lower()
def sum_list(list):
  s=0
  for i in list:
    s+=i
  return s
while play=='yes':
  players_cards=[]
  players_cards.append(random.choice(deck))
  players_cards.append(random.choice(deck))
  computers_cards=[]
  computers_cards.append(random.choice(deck))
  print("Your cards are:",players_cards)
  print("Computers first card:",computers_cards)
  more=input("Do you want another card? ").lower()
  if more=="yes":
    players_cards.append(random.choice(deck))
    print("Your final cards are : ",players_cards)
  else:
    print("Your final cards are :",players_cards)
  computers_cards.append(random.choice(deck))
  print("The computer's final cards are :",computers_cards)
  while sum_list(players_cards)>21:
    if 11 in players_cards:
      players_cards[players_cards.index(11)]=1
    else:
      print(You_lose)
  if sum_list(players_cards)<sum_list(computers_cards):
    print(You_lose)
  else:
    print(You_won)
  play=input("Do you want to play BlackJack??? ").lower()
```
